```python
class student:
    
    # default constructure
     def __init__(self):
        pass

    # parameterized constructure 
     def __init__(self,name,marks):
        self.name= name
        self.marks= marks
        print("Adding new student in database.")

s1=student("Atul Gangwar",98)
print(s1.name,s1.marks)

s2=student("Kunal",96)
print(s2.name,s2.marks)
```

# *args

```python
# *args

def my_function(*args):
  print("Type:", type(args))
  print("First argument:", args[0])
  print("Second argument:", args[1])
  print("All arguments:", args)

my_function("Atul", "kunal", "Ritik","Tiger","Bihari")
```

# **kwargs

```python
# **kwargs

def my_function(**kwargs):
  print("Type:", type(kwargs))
  print("Name:", kwargs["name"])
  print("Age:", kwargs["age"])
  print("All data:", kwargs)

my_function(name = "Atul Gangwar", age = 18, city = "Bareilly")
```

# Methods

```python

class Student:
    college_name="Invertis University"      # Class Attribute
    
    def __init__(self,name,marks):
            self.name= name          # Instance/object Attribute
            self.marks= marks
        
    def welcome(self):
        print("Welcome ",self.name)

    def get_marks(self):
        return self.marks
        
s1=Student("Atul Gangwar",98)
print(s1.name,s1.marks)
s1.welcome()
s1.get_marks()
```

```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def average(self):
        sum=0
        for val in self.marks:
            sum += val
        print("Hi",self.name,"Your average score is ",sum/3)

s1 = Student("Atul",[94,96,98])
s1.average()
        
```
