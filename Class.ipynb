```python
name=input("Enter your name : ")
age=int(input("Enter youyr age:"))
mark1, mark2, mark3 = map(float, input("Enter marks in 3 subjects (separated by space): ").split())

total = mark1 + mark2 + mark3
average = total / 3

print("\n--- Student Report ---")
print(f"Name: {name}")
print(f"Age: {age}")
print("Marks:", mark1, mark2, mark3, sep=",")
print("Total Marks =", total)
print("Average Marks =", round(average, 2))

print("Thank You", end=" ")
print("for using this program!")
```

```python
# using while loop
i=5
while i>=1:
    print("* "*i)
    i-=1
```

```python
# palindrome check using slicing + operator

num=input("Enter a Integer who you want to check for palindrome : ")
rev=num[::-1]

if num==rev:
    print("Reverse of your value is : ",rev,"\nSo Your value is palindrome.")
else:
    print("Reverse of your value is : ",rev,"\nSo Your value is not palindrome.")
```
