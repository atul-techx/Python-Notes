# 4. Rows & Columns - Selection

```python
# Select Rows
# loc - Label based index
Load_df.loc[Load_df.Name=='Atul']
```

```python
Load_df.loc[(Load_df.Name=='Aditya')&(Load_df.M_salary>=50000)]
```

```python
Load_df.loc[0:2] # include 0-2 all value 
```

```python
# iloc - index-value based
Load_df.iloc[2]
```

```python
# Multiple rows
Load_df.iloc[0:2]    # [start:stop:step] include value accept 2
```
