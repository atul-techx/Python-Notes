```python
from collections import deque  
# here is collections is a library & deque is a module
```

```python
checkout = deque()
checkout.append("Cus1")
checkout.append("Cus2")
checkout.append("Cus3")
checkout.append("Cus4")
checkout
```

```python
while checkout:
    customer = checkout.popleft()
    print("serving",customer)
```

```python
from queue import Queue
print_queue = Queue()
print_queue.put("Print the 1 pdf")   # put is similar to push
print_queue.put("Print the 2 pdf")
print_queue.put("Print the 3 pdf")
print_queue.put("Print the 4 pdf")
print_queue.put("Print the 5 pdf")

while not print_queue.empty():
    print_job = print_queue.get()  # get is similar to pop
    print("Printing",print_job)
    
```

```python
# write a function that takes list as input and return all numeric values

def only_numeric(a):
    n=[]
    for i in a:
        if type(i)==int or type(i)==float:
            n.append(i)
    return n

only_numeric([1,True,2,"Hello",4,"Atul",2.5,3+5j])
```

```python
# if we want to get numeric value inside a nested list

def only_numeric(a):
    n=[]
    for i in a:
        if type(i)==list:
            for j in i:
                if type(j)==int or type(j)==float:
                    n.append(j)
        else:
            if type(i)==int or type(i)==float:
                n.append(i)
    return n

only_numeric([[5.9,"Ankit",3+8j],1,True,2,"Hello",4,"Atul",2.5,3+5j,[15,23,"Data"]])
```

# generator function

```python
# Regular function calculates the square of the values and return in one go
# Since its returning in one go >> memory usage will be more and 
# it will a lot of execution time if there is a lot of computation as all the computations
# will be done first and then the calculation will be returned in one go

def square_number(n):
    result = []
    for i in range(n):
        result.append(i**2 )
    return result
square_number(10)
```

```python
# can we have somthing to generate result one by one instead of one go
# yield >> general function uses return statement but a generator function uses yield

def square_number_generator(n):
    for i in range(n):
        yield i**2
```

```python
square_number_generator(10)  # yield is creating a generator similar to object
```

```python
gen = square_number_generator(10)
gen
```

```python
# fibonacci series >> sum of last two numbers 0,1,1,2,3,5,8,13...

def fib(n):
    a = 0
    b = 1
    for i in range(n):  # if n=5 , 0,1,2,3,4
        yield a
        a, b = b, a+b

fib(1000)
```

```python
# Way to provide an option to see the private method >> a wrapper
class Student:
    def __init__(self, name, degree):
        self.name = name
        self.__degree = degree  # Private Variable
        
    def show(self):
        # accessing the private variable
        print("name", self.name, "Degree", self.__degree)
    
    def __Private_method(self):
        print("this is a private method")
        
    def access_Private_method(self):
        self.__Private_method()
```

```python
std = Student("Atul","BCA")
```

```python
std.access_Private_method()
```

```python
# Another Use case

class Bank:
    def __init__(self, balance):
        self.__balance = balance

    def Deposite(self, amount):
        self.__balance = self.__balance + amount

    def Withdraw(self, amount):
        if self.__balance >= amount:
            self.__balance = self.__balance - amount
            return True
        else:
            return False
    def get_balance(self):
        return self.__balance
```

```python
# Protected >> within the class and its subclass, protected member can be accessed, writing with ( _ )
```

```python
class College:
    def __init__(self):
        self._college_name = "Invertis University"
class Student(College):
    def __init__(self, name):
        self.name = name
        College.__init__(self)   # accessing variable of base class

    def show(self):
        print("Name:", self.name, "College:", self._college_name)  # directly call the Base class variable
        
```

```python
std = Student("Atul")
```

```python
std.name
```

```python
std.show()
```

```python
coll = College()
coll._college_name  # accessing the protected variable
```

```python
# Another way to accessing the data of Base class using super()

class College:
    def __init__(self):
        self._college_name = "Invertis University"
class Student(College):
    def __init__(self, name):
        self.name = name
        super().__init__()   # include super and remove self, base class is super class

    def show(self):
        print("Name:", self.name, "College:", self._college_name)  # directly call the Base class variable
```

```python
std = Student("Atul")
```

```python
std.show()
```

```python
class Student:
    def __init__(self, name):
        self.name = name
    @classmethod   # bound/binds to the class
    def student_details(cls, name):   # instead of self as first parameter you are using cls meaning this method is associated/refering to the class itself
        return cls(name)
```

```python
# What is the meaning of that it is referring to the class ?
# earlier to access any method/attribute >> first make instance/object of class
# then obj.attribute

# But if it is class method you can directly access it
```

```python
obj = Student.student_details("Atul") # since student_details directly binds to the class you can directly call it without making any object
```

```python
obj.name
```

```python
# How class method is overloading the init method ?
# Modifying the instance method 
class Student:
    def __init__(self, name):  # init method is instance method
        self.name = name
        
    @classmethod   
    def student_details(cls, name1):   
        return cls(name1)
```

```python
std = Student.student_details("AtulG")
```

```python
std.name
# you will get AtulG inspite the variable name1
# because it is overloading the init method >> alternative of init method
```

```python
# class method can modify class state applicable to all the instances of the class
class Student:

    total_students = 0  # class variable 
    def __init__(self, name):  # instance method
        self.name = name     # instance variable
        Student.total_students = Student.total_students + 1   # total student is the class variable associated with all the instances

    @classmethod
    def get_total_students(cls):
        return cls.total_students
```

```python
Student.total_students
```

```python
std = Student("Atul")
std1 = Student("Vijay")
```

```python
std.name
```

```python
std1.name
```

```python
Student.total_students
```

```python
Student.get_total_students()  # class method can modify class state applicable to all the instances of the class
# thats why you see increase in the count
```

```python
# To make external functions as class method

class Student:

    total_students = 0  # class variable 
    def __init__(self, name):  # instance method
        self.name = name     # instance variable
        Student.total_students = Student.total_students + 1   # total student is the class variable associated with all the instances

    @classmethod
    def get_total_students(cls):
        return cls.total_students
```

```python
def course_details(cls, course_name):
    print("The details of the course are :", course_name)
    
```

```python
Student.course_details = classmethod(course_details)  # adding external function as class
```

```python
Student.course_details("Data science")
```

```python
# Delete any class method

del Student.course_details
```
